```python
import pyspark
```

```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName('Practise').getOrCreate()
```

```python
sc = spark.sparkContext
```

```python
text_file = sc.textFile('test1.csv')
```

```python
avg_score = (text_file.map(lambda line: line.strip().split(','))
             .filter(lambda l: len(l)==3)
             .map(lambda l: (l[0], (int(l[2]),1)))
             .reduceByKey(lambda x,y: (x[0]+y[0], x[1]+y[1]))
             .map(lambda t: (t[0], t[1][0]/t[1][1])))
print(avg_score.collect())
             
```
